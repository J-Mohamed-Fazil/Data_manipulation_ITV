```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv('Cleaned.csv')
print("Dataset Loaded Successful")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Profit_Amount']= (df['Revenue']*df['Profit_Margin']/100)
```

```python
df['Profit_Amount']
```

```python
df[["Revenue",
    "Profit_Margin",
    "Profit_Amount"]].sample(5)
```

```python
df['Stock_Difference'] = (
    df['Quantity_Available'] - df['Units_Sold']
)
```

```python
	
df[["Quantity_Available",
    "Units_Sold",
    "Stock_Difference"]].sample(5)
```

```python
df['Inventory_Status'] = np.where(
    df['Quantity_Available']<=200,
    'Low Stock',
    np.where(
        df['Quantity_Available']<=200,
        'Medium Stock',
        'High Stock'
    )
)
```

```python
	
df[["Quantity_Available",
    "Inventory_Status"]].sample(5)
```

```python
df["Revenue_Per_Unit"] = (
    df["Revenue"] /
    df["Units_Sold"]
)

df[["Revenue",
    "Units_Sold",
    "Revenue_Per_Unit"]].sample(5)
```

```python
print(df[[
    "Profit_Amount",
    "Stock_Difference",
    "Inventory_Status",
    "Revenue_Per_Unit"
]].head())
```

```python
df["Revenue_Category"] = np.where(
    df["Revenue"] < 5000,
    "Low Revenue",
    np.where(
        df["Revenue"] < 10000,
        "Medium Revenue",
        "High Revenue"
    )
)

df[["Revenue", "Revenue_Category"]].head()
```

```python
df["Customer_Feedback"] = np.where(
    df["Customer_Satisfaction"] >= 4,
    "Satisfied",
    "Needs Improvement"
)

df[[
    "Customer_Satisfaction",
    "Customer_Feedback"
]].head()
```

```python
df["Discount_Category"] = np.where(
    df["Max_Discount_Offered"] < 10,
    "Low Discount",
    np.where(
        df["Max_Discount_Offered"] < 20,
        "Medium Discount",
        "High Discount"
    )
)

df[["Max_Discount_Offered", "Discount_Category"]].sample(5)
```

```python
df["Delivery_Speed"] = np.where(
    df["Delivery_Time"] <= 5,
    "Fast Delivery",
    "Standard Delivery"
)

df[[
    "Delivery_Time",
    "Delivery_Speed"
]].head()
```

```python
print(df[[
    "Revenue_Category",
    "Customer_Feedback",
    "Discount_Category",
    "Delivery_Speed"
]].head())
```
